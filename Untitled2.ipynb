```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

def distance(p1, p2):
    return (p1 - p2) ** 2

def kmeans_apply(img, k, iterations=10):
    img_rows, img_cols = img.shape
    cluster_centers = np.random.choice(img.flatten(), k, replace=False).astype(np.float32)
    labels = np.zeros((img_rows, img_cols), dtype=np.uint8)

    for _ in range(iterations):
        for i in range(img_rows):
            for j in range(img_cols):
                pixel_value = img[i, j]
                distances = [distance(pixel_value, center) for center in cluster_centers]
                labels[i, j] = np.argmin(distances)

        new_centers = np.zeros_like(cluster_centers)
        counts = np.zeros(k, dtype=np.int32)

        for i in range(img_rows):
            for j in range(img_cols):
                cluster_idx = labels[i, j]
                new_centers[cluster_idx] += img[i, j]
                counts[cluster_idx] += 1

        for i in range(k):
            if counts[i] > 0:
                new_centers[i] /= counts[i]

        if np.all(cluster_centers == new_centers):
            break

        cluster_centers = new_centers.copy()

    clustered_image = (255 / (k - 1)) * labels
    clustered_image = clustered_image.astype(np.uint8)
    return clustered_image, cluster_centers, labels

def compute_Q(img, labels, centers):
    Q = 0
    img_rows, img_cols = img.shape
    for i in range(img_rows):
        for j in range(img_cols):
            cluster_idx = labels[i, j]
            Q += distance(img[i, j], centers[cluster_idx])
    return Q

img = cv2.imread('/content/Objects.bmp', 0)
k_values = range(2, 11)
Q_values = []

for k in k_values:
    segmented_img, centers, labels = kmeans_apply(img, k)
    Q = compute_Q(img, labels, centers)
    Q_values.append(Q)

plt.figure()
plt.plot(k_values, Q_values, marker='o', linestyle='-')
plt.xlabel("Number of Clusters (K)")
plt.show()
best_k = k_values[np.argmin(Q_values)]
print(f"Best K: {best_k}")
segmented_img, _, _ = kmeans_apply(img, best_k)
cv2_imshow(segmented_img)
plt.title(f"Best Segmentation (K={best_k})")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

x = np.array([[1, 1], [1, 2], [2, 1], [0, 10], [10, 11], [11, 10]])
kmeans = KMeans(n_clusters=2, random_state=0, n_init=10).fit(x)
plt.scatter(x[:, 0], x[:, 1], c=kmeans.labels_, cmap='viridis', s=100, edgecolors='k')
plt.scatter(*kmeans.cluster_centers_.T, c='red', marker='X', s=200, label='Centers')

plt.title("(K=2)")
plt.legend()
plt.show()
```
